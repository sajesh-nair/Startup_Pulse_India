The Inspection & Cleaning Script

```python
import pandas as pd

# 1. Ingest the actual registry data
df = pd.read_csv("data/dpiit_startups_2026.csv")

# 2. Clean column headers
df.columns = [col.strip().replace(" ", "") for col in df.columns]

# 3. Clean text values inside columns to avoid hidden spacing bugs
df['state'] = df['state'].astype(str).str.strip()
df['industry'] = df['industry'].astype(str).str.strip()

print("✅ Data Loaded and Cleaned Successfully!")
print(f"Total rows: {len(df)}")
```

Inspecting Unique Sectors & States

```python
# Print unique count
print(f"Total unique industries: {df['industry'].nunique()}")
print(f"Total unique states: {df['state'].nunique()}")

print("\n--- Top 15 Industries by Row Count ---")
print(df['industry'].value_counts().head(15))

print("\n--- Unique States Listed ---")
print(sorted(df['state'].unique()))
```

Top 10 Industries Analysis (Bar Chart)

```python
import plotly.express as px

# 1. Group by industry and sum the recognized startups
industry_totals = df.groupby("industry")["startups_recognized"].sum().reset_index()

# 2. Sort to find the true top 10 heavyweights
top_10_industries = industry_totals.sort_values(by="startups_recognized", ascending=False).head(10)

# 3. Create a clean, professional horizontal bar chart
fig_industry = px.bar(
    top_10_industries,
    x="startups_recognized",
    y="industry",
    orientation="h",
    color="startups_recognized",
    color_continuous_scale="Viridis",
    title="Top 10 Indian Industries by Total DPIIT Recognized Startups",
    labels={"startups_recognized": "Total Startups Registered", "industry": "Industry Sector"},
    template="plotly_dark"
)

# Improve layout presentation
fig_industry.update_layout(yaxis={'categoryorder':'total ascending'}, showlegend=False)
fig_industry.show()
```

Time-Series Analysis (Year-on-Year Growth)

```python
# 1. Define the top 5 industries we want to track based on your bar chart
top_5_sectors = ['IT Services', 'Healthcare & Lifesciences', 'Education', 'Construction', 'Finance Technology']

# 2. Filter the dataframe for only these 5 sectors
df_top_5 = df[df['industry'].isin(top_5_sectors)]

# 3. Group by Year and Industry to get the registration trends
trend_data = df_top_5.groupby(['year', 'industry'])['startups_recognized'].sum().reset_index()

# 4. Plot an interactive line chart
fig_line = px.line(
    trend_data,
    x="year",
    y="startups_recognized",
    color="industry",
    markers=True,
    title="Year-on-Year Registration Trajectory (Top 5 Sectors)",
    labels={"startups_recognized": "New Startups Recognized", "year": "Calendar Year"},
    template="plotly_dark"
)

fig_line.update_layout(hovermode="x unified")
fig_line.show()
```

```python
# 1. Group by state and sum the recognized startups
state_totals = df.groupby("state")["startups_recognized"].sum().reset_index()

# 2. Grab the top 10 states by volume
top_10_states = state_totals.sort_values(by="startups_recognized", ascending=False).head(10)

# 3. Plot a gorgeous horizontal bar chart
fig_state = px.bar(
    top_10_states,
    x="startups_recognized",
    y="state",
    orientation="h",
    color="startups_recognized",
    color_continuous_scale="Cividis",
    title="Top 10 Indian States by DPIIT Startup Registrations",
    labels={"startups_recognized": "Total Startups Registered", "state": "State / UT"},
    template="plotly_dark"
)

# Reverse the y-axis so the highest state is at the top
fig_state.update_layout(yaxis={'categoryorder':'total ascending'}, showlegend=False)
fig_state.show()
```
